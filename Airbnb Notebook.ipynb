#Using the output above we will create a list of columns to keep and columns to drop

```python
Columns_to_keep = ['NAME', 'host id', 'host_identity_verified', 'host name',
       'neighbourhood group', 'neighbourhood', 'lat', 'long', 'country',
       'country code', 'instant_bookable', 'cancellation_policy', 'room type',
       'Construction year', 'price', 'service fee', 'minimum nights',
       'number of reviews']
Columns_to_drop = ['id', 'last review', 'reviews per month',
       'review rate number', 'calculated host listings count',
       'availability 365', 'house_rules', 'license']
```

#Filtering our dataset to show only relevant columns

```python
print(Data[Columns_to_keep])
```

#We can clean our data by uisng the lists we created to drop(delete) the redundant columns. This is a permanent action unlike what we did before with the filter.

```python
Data = Data.drop(columns= Columns_to_drop)
print(Data)
```

#Creating a standard column Title format for our dataset.

#Handling NaN(null) values in the dataset based on the importance and type of each column.

```python
print(Data.isna().sum()) #Loads the sum of null values in each column
```

```python
"""
Drop rows only where very important values are missing:
Price, Latitude, or Longitude are important for cost and location analysis. 
The number of null rows for lat and long is too small to hurt when dropped
We can keep or fill null price rows if we dont need to do a cost analysis or the number of rows missing is small,
in this case it's better to drop the price null rows as it will hinder visualization and comparing listings by price.
"""
Data.dropna(subset=["Price", "Latitude", "Longitude"])
```

```python
Data.columns
```

```python
#dropna failed because the columns are stored as a multi index so i will flatten them to remove the extra column level and make column names normal
Data.columns = Data.columns.get_level_values(0) 
Data.dropna(subset=["Price", "Latitude", "Longitude"], inplace=True) 
```

```python
# Fill numeric columns with median
numeric_cols = ["Service_Fee", "Minimum_Nights", "Construction_Year", "Number_Of_Reviews"]
for col in numeric_cols:
    Data[col] = Data[col].fillna(Data[col].median())
```

```python
"""
The code above throws an error because the service fee column has a dollar sign  and that makes the column a string.
i will fix this then reload the code.
"""
# Remove $ and convert to numeric, the 'errors' is incase the column has missing values or bad text
Data["Service_Fee"] = pd.to_numeric(Data["Service_Fee"].replace("[$,]", "", regex=True), errors="coerce")  
numeric_cols = ["Service_Fee", "Minimum_Nights", "Construction_Year", "Number_Of_Reviews"]
for col in numeric_cols:
    Data[col] = Data[col].fillna(Data[col].median())
```

```python
# Fill text columns with "Unknown"
Data["Host_Name"] = Data["Host_Name"].fillna("Unknown")
Data["Country"] = Data["Country"].fillna("Unknown")
Data["Name"] = Data["Name"].fillna("Unknown")
Data["Neighbourhood_Group"] = Data["Neighbourhood_Group"].fillna('Unknown')
Data["Neighbourhood"] = Data["Neighbourhood"].fillna("Unknown")
Data["Country_Code"] = Data["Country_Code"].fillna('unknown')
```

```python
# Fill remaining categorical columns with mode
categorical_cols = ["Host_Identity_Verified", "Instant_Bookable", "Cancellation_Policy"]

for col in categorical_cols:
    Data[col] = Data[col].fillna(Data[col].mode()[0])
```

```python
Data.isna().sum() #Rechecking the sum of the null values across columns to see if the changes reflected
```

#Cleaning Individual Columns

```python
# changing the values in Host_identity_verified to uppercase letters
Data['Host_Identity_Verified'] = Data['Host_Identity_Verified'].str.upper()
```

```python
Data['Host_Identity_Verified']
```

```python
#The price column has the dollar sign, which makes it a string value and will make price analysis almost impossible, so I will fix that.
Data["Price"] = pd.to_numeric(Data["Price"].replace("[$,]", "", regex=True), errors="coerce") 
Data['Price']
```

```python
#The construction year is a float value with decimal, I think its best to convert it to an integer
Data["Construction_Year"] = Data["Construction_Year"].astype(int)
Data["Construction_Year"]
```

```python
Data['Neighbourhood_Group'].unique()
"""
There's 8 distinct values in the Neighbourhood_Group column instead of 6 because "Brooklyn" and "Manhattan" were spelled wrong in some rows. 
I will be fixing this so it doens't affect our analysis
"""
Data["Neighbourhood_Group"] = Data["Neighbourhood_Group"].replace({
    "brookln": "Brooklyn",
    "manhatan": "Manhattan"
})  # Correct misspelled neighbourhood names
```

```python
Data['Neighbourhood_Group'].unique() #rechecking to confirm our changes were effected
```

```python
Data['Room_Type'].unique()
```

```python
"""We have some rows with less than 0 minimum night, negative values in Minimum_Nights are not logically valid,
so they are likely data entry errors and should be cleaned
"""
Data[Data["Minimum_Nights"] < 0]
```

```python
(Data["Minimum_Nights"] < 0).sum() # Count rows where Minimum_Nights is less than zero
```

```python
Data = Data[Data["Minimum_Nights"] >= 0]  # Remove rows with negative minimum nights
```
